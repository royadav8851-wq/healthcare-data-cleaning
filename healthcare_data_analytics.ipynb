```python
import pandas as pd

print("Pandas is ready!")
```

```python
data = {
    "Patient_ID": ["P001", "P002", "P003", "P004", "P005",
                   "P006", "P007", "P008", "P009", "P010"],
    "Age": [25, 45, 32, 28, 60, 25, 40, 35, 50, 22],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Male", "Female", "Male", "Female", "Male"],
    "Disease": ["Flu", "Hypertension", "Diabetes", "Asthma",
                "Arthritis", "Flu", "Diabetes", "Flu",
                "Hypertension", "Asthma"],
    "Medication": ["Paracetamol", "Amlodipine", "Metformin",
                   "Salbutamol", "Ibuprofen", "Paracetamol",
                   "Metformin", "Paracetamol", "Amlodipine",
                   "Salbutamol"],
    "Dosage_mg": [500, 5, 500, 100, 400, 500, 500, 500, 5, 100]
}

df = pd.DataFrame(data)

display(df)
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df.loc[2, "Age"] = None

print(df.isnull().sum())
```

```python
average_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(average_age)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate records:", df.duplicated().sum())
```

```python
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("Rows after adding duplicate:", len(df))
print("Duplicate records:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Rows after cleaning:", len(df))
print("Duplicates remaining:", df.duplicated().sum())
```

```python
patient_count = df["Patient_ID"].nunique()

print("Total unique patients:", patient_count)
```

```python
disease_count = df["Disease"].value_counts()

print("Disease frequency:")
print(disease_count)

print("\nMost common disease:")
print(disease_count.idxmax())
```

```python
average_age = df["Age"].mean()

print("Average patient age:", round(average_age, 2), "years")
```

```python
print("===== HEALTHCARE DATA ANALYSIS =====")

print("Total unique patients:", df["Patient_ID"].nunique())

print("\nDisease frequency:")
print(df["Disease"].value_counts())

print("\nMost common disease:", df["Disease"].value_counts().idxmax())

print("\nAverage patient age:", round(df["Age"].mean(), 2), "years")

print("\nMissing values remaining:", df.isnull().sum().sum())

print("Duplicate records remaining:", df.duplicated().sum())
```
